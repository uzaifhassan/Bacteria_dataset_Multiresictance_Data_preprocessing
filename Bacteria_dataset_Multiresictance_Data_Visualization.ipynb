# Data Visualization 

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
df=pd.read_csv("Update_Bacteria_dataset.csv")
df.head(20)
```

```python
df.columns
```

```python
df.info()
```

## Which bacterial strain (Souches) appears most frequently in the dataset?

```python
df["Souches"].value_counts()
```

```python
top = df["Souches"].value_counts().head(300)

plt.figure(figsize=(20,100))

sns.barplot(
    y=top.index,
    x=top.values
)

plt.xlabel("Count")
plt.ylabel("Souches")
plt.show()
```

## What is the distribution of patients by gender?

```python
plt.figure(figsize=(10,8))
tot=df["Gender"].value_counts()
tot.plot(kind="bar")

plt.xlabel("Gender")
plt.ylabel("Numbers Patients")
plt.show()
```

## How are patients distributed across different age groups?

```python
plt.figure(figsize=(10,8))
bins = [0, 20, 40, 60, 80, 100]
labels = ["0-20", "21-40", "41-60", "61-80", "81-100"]
df["Age_Group"] = pd.cut(df["Age"], bins=bins, labels=labels)
patient_grp=df["Age_Group"].value_counts()
plt.pie(patient_grp,autopct="%1.1f%%",labels=labels)

plt.xticks(rotation=45)
plt.show()
```

## Which antibiotic has the highest number of resistant cases among all antibiotics?

```python
plt.figure(figsize=(12,10))
rows=df[[
    'AMX/AMP', 
    'AMC', 
    'CZ', 
    'FOX', 
    'CTX/CRO', 
    'IPM', 
    'GEN', 
    'AN',
    'ofx', 
    'CIP', 
    'C'
    ]]

total=[]
for i in rows:
    st=rows[i].value_counts()
    total.append(st["R"])
print(total)

sns.barplot(x=rows.columns, y=total)
plt.ylabel("resistant cases")
plt.xlabel("Antibiotic")
plt.xticks(rotation=45)
plt.show()
```

## Is there any relationship between patients' age and infection frequency (Infection_Freq)?

```python
df.columns
```

```python
plt.figure(figsize=(11,10))
plt.scatter(df["Age"],df["Infection_Freq"])
plt.xlabel("Age")
plt.ylabel("Infection_Freq")
plt.xticks(rotation=45)
plt.show()
```

## How does infection frequency differ between diabetic and non-diabetic patients?

```python
plt.figure(figsize=(8,6))


ax=sns.barplot(x="Diabetes",y="Infection_Freq",data=df)
plt.xlabel("Diabetes")
plt.ylabel("Infection_Freq")
plt.legend(title="Diabetes")
plt.xticks(rotation=45)
plt.legend(labels=["Male", "Female"])
plt.show()
```

## What is the comparison of hypertension status among male and female patients?

```python
plt.figure(figsize=(8,6))

sns.countplot(x="Gender", hue="Hypertension", data=df)
plt.xlabel("Gender")
plt.ylabel("Number of Patients")
plt.legend(title="Gender")
plt.xticks(rotation=45)
plt.legend(title="Hypertension")
plt.show()
```

## How has the number of bacterial samples (Collection_Date) changed over time?

```python
plt.figure(figsize=(8,6))
df["Date"]=pd.to_datetime(df["Collection_Date"],errors='coerce')
dt=df["Date"].value_counts()
v1=dt.values
print(dt)
sns.lineplot(x=dt.index,y=v1)
plt.xlabel("Collection Date")
plt.ylabel("Number of bacterial")
plt.xticks(rotation=45)
plt.show()
```

### Which bacterial strains (Souches) are most commonly resistant to the antibiotic CIP?

```python
d1=df[df["CIP"]=="R"][["Souches","CIP"]]
ct=d1.value_counts()
sh = d1["Souches"].value_counts().head(500)
print(sh)
plt.figure(figsize=(6,100))

sns.barplot(y=sh.index, x=sh.values)
plt.xlabel("Bacterial Strains (Souches)")
plt.ylabel("Number of Resistant Cases (CIP = R)")
plt.xticks(rotation=45)
plt.show()
```
