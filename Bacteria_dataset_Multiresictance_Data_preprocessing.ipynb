```python
cols = ["FOX", "CTX/CRO", "IPM", "GEN", "AN"]

df[cols] = (
    df[cols]
    .apply(lambda x: x.str.strip() if x.dtype == "object" else x)
    .replace({
        "s": "S",
        "r": "R",
        "i": "I",
        "Intermediate": "I",
        "?": np.nan,
        "missing": np.nan
    })
)
```
