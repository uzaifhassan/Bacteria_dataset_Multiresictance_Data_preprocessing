```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

df=pd.read_csv("Bacteria_dataset_Multiresictance.csv")
df.head(10)
```

```python
df.tail()
```

```python
missing_value=round((df.isnull().sum())/len(df)*100,2)
print(missing_value)
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.columns
```

```python
df["age/gender"].fillna(df["age/gender"].mode()[0],inplace=True)
```

```python
df["Souches"].fillna(df["Souches"].mode()[0],inplace=True)
```

```python
df[["Age","Gender"]]=df['age/gender'].str.split('/',expand=True)
```

```python
df.head(10)
```

```python
df["Age"] = pd.to_numeric(df["Age"], errors="coerce")
```

```python
df["Gender"].isnull().sum()
```

```python
df["Age"].fillna(df["Age"].mean(),inplace=True)
```

```python
Q1 = df["Age"].quantile(0.25)
Q3 = df["Age"].quantile(0.75)
IQR = Q3 - Q1

outliers = df[(df["Age"] < (Q1 - 1.5 * IQR)) |
              (df["Age"] > (Q3 + 1.5 * IQR))]

print("Total Outliers:", len(outliers))
print(outliers)
```

```python
df.drop("age/gender",axis=1,inplace=True)
```

```python
email_index = df.columns.get_loc("Email")

df.insert(email_index + 1, "Age", df.pop("Age"))
df.insert(email_index + 2, "Gender", df.pop("Gender"))
```

```python
invalid_gender = df[~df["Gender"].isin(["M", "F"])]
print(invalid_gender)
print("Invalid Gender:", len(invalid_gender))
```

```python
df.isnull().sum()
```

```python
df["Souches"].value_counts()
```

```python
df["Souches"]=df["Souches"].replace("?",np.nan)
```

```python
store=["S290 Escherichia coli",
"S291 Morganella morganii",
"S292 Proteus mirabilis",
"S10287 Morganella morganii",
"S10289 Escherichia coli",
"S10288 Escherichia coli",
]

mask=df["Souches"].isna()
df.loc[mask, "Souches"] = np.random.choice(store, size=mask.sum())
```

```python
df["Diabetes"].value_counts()
```

```python
missing=round((df["Diabetes"].isnull().sum()/len(df))*100,2)
print(f"percentage of missing value:{missing:.2f}%")
```

```python
df["Diabetes"]=df["Diabetes"].replace("missing",np.nan)
```

```python
df["Diabetes"].fillna(df["Diabetes"].mode()[0],inplace=True)
```

```python
df["Hypertension"].value_counts()
```

```python
df["Hypertension"]=df["Hypertension"].replace("?",np.nan)
```

```python
df["Hypertension"].fillna(df["Hypertension"].mode()[0],inplace=True)
```

```python
df.info()
```

```python
df["Age"]=df["Age"].astype(int)
```
